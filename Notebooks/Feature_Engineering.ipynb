### Feature Engineering is the process of creating meaningful features from existing data to improve analysis and decision-making. In this project, key features such as Risk Level, Battery Status, Temperature Status, Date, Hour, and Route were derived from the raw IoT sensor data. These engineered features help identify abnormal container conditions, monitor shipment health, enable time-based analysis, and provide valuable insights for SQL analysis and Power BI dashboard visualization.

## Import Libraries

```python
import pandas as pd
import numpy as np
```

## Load Dataset

```python
df = pd.read_csv(r"C:\Users\sayam\Desktop\Atmosync\Data\sensor_data.csv")
df.head()
```

## Risk Level

```python
def risk_level(row):
    if (
        row['temperature'] > 8 or
        row['humidity'] > 90 or
        row['vibration'] > 4 or
        row['battery_level'] < 20
    ):
        return "High"
    elif (
        row['temperature'] > 6 or
        row['humidity'] > 85 or
        row['vibration'] > 3 or
        row['battery_level'] < 40
    ):
        return "Medium"

    return "Low"

df['risk_level'] = df.apply(risk_level, axis=1)
```

## Battery Status

```python
def battery_status(x):
    if x >= 80:
        return "Good"
    elif x >= 40:
        return "Moderate"
    else:
        return "Low"
df['battery_status'] = df['battery_level'].apply(battery_status)
```

## Temperature Status

```python
def temperature_status(x):
    if x < 2:
        return "Low"
    elif x <= 8:
        return "Normal"
    else:
        return "High"
df['temperature_status'] = df['temperature'].apply(temperature_status)
```

## Timestamp Features

```python
df['timestamp'] = pd.to_datetime(df['timestamp'])
df['date'] = df['timestamp'].dt.date
df['hour'] = df['timestamp'].dt.hour
```

## Shipment route

```python
df['route'] = df['origin'] + " → " + df['destination']
```

```python
# Save the feature-engineered dataset
df.to_csv("sensor_data_feature_engineered.csv", index=False)
```
