## Import Required Libraries

```python
import pandas as pd
import random
from datetime import datetime, timedelta
import warnings
warnings.filterwarnings("ignore")
```

```python
# Number of records to generate
num_records = 10000

# Create 100 container IDs
containers = [f"C{str(i).zfill(3)}" for i in range(1, 101)]

# Create 1000 shipment IDs
shipments = [f"S{str(i).zfill(5)}" for i in range(1,1001)]

# fruit types transported in refrigerated containers
fruits = [
    "Apple",
    "Banana",
    "Mango",
    "Orange",
    "Grapes",
    "Pomegranate",
    "Strawberry",
    "Kiwi",
    "Papaya",
    "Pineapple"
]

# cities used as shipment origin and destination
cities = [
    "Mumbai",
    "Delhi",
    "Bangalore",
    "Hyderabad",
    "Chennai",
    "Kolkata",
    "Pune",
    "Ahmedabad"
]
#coordinates of cities
city_coordinates = {
    "Mumbai": (19.0760, 72.8777),
    "Delhi": (28.6139, 77.2090),
    "Bangalore": (12.9716, 77.5946),
    "Hyderabad": (17.3850, 78.4867),
    "Chennai": (13.0827, 80.2707),
    "Kolkata": (22.5726, 88.3639),
    "Pune": (18.5204, 73.8567),
    "Ahmedabad": (23.0225, 72.5714)
}

# empty list to store all generated sensor data
rows=[]
```

## Generate historical IoT sensor records

```python
for i in range(num_records):
    #select a random city
    origin=random.choice(cities)

    #select a destination city
    destination= random.choice([city for city in cities if city != origin])

    #generate one sensor record
    record = {"container_id": random.choice(containers),
              "shipment_id": random.choice(shipments),
              "fruit_type": random.choice(fruits),
              "origin": origin,
              "destination": destination,

        # Temperature inside refrigerated container (°C)
        "temperature": round(random.uniform(2, 8), 2),

        # Humidity inside container (%)
        "humidity": round(random.uniform(70, 95), 2),

        # Vibration level during transportation
        "vibration": round(random.uniform(0, 5), 2),

        # Battery percentage of IoT device
        "battery_level": random.randint(70, 100),

        # Door status (mostly closed)
        "door_status": random.choices(
            ["Closed", "Open"],
            weights=[95, 5]
        )[0],

        # Current status of the container
        "container_status": random.choices(["In Transit", "Loading", "Unloading", "Delivered"],weights=[60,15,10,15])[0],
              
        # GPS location near the origin city
        "latitude": round(city_coordinates[origin][0] + random.uniform(-0.05,0.05),6),

        "longitude": round(city_coordinates[origin][1] + random.uniform(-0.05,0.05),6),

        # Historical timestamp
        "timestamp": (
            datetime.now() -
            timedelta(minutes=random.randint(0, 10080))
        ).strftime("%Y-%m-%d %H:%M:%S")
    }

    # Store the generated record
    rows.append(record)
```

## Create dataframe and save csv

```python
df = pd.DataFrame(rows)
df.to_csv("../Data/sensor_data.csv", index=False)
print("Dataset generated successfully!")
```
